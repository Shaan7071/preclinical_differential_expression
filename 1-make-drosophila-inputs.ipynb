```python
import os
import pandas as pd 
import re
```

```python
# Config
DATASET = 'Drosophila-2025' 
RAW_DIR = os.path.join(DATASET, 'raw-data')
PROCESSED_DIR = os.path.join(DATASET, 'processed-data')
DEA_DIR = os.path.join(DATASET, "DEA-results")
```

**Counts Matrix**

```python
counts_path = os.path.join(RAW_DIR, "deseq_counts.csv")
counts_df = pd.read_csv(counts_path, index_col=0)
counts_df
counts_df.shape
# counts are already in samples x genes format for pyDESeq2
counts_df.columns.name = None
counts_df.index.name = "sample"
# Save processed counts to a new file 
counts_df = counts_df.reset_index()
processed_counts_path = os.path.join(PROCESSED_DIR, "counts.csv")
counts_df.to_csv(processed_counts_path, index=False)
```

**Metadata**

```python
metadata_in = os.path.join(RAW_DIR, "deseq_metadata.csv")
metadata_df = pd.read_csv(metadata_in, index_col=0)
metadata_df
# metadata is already good 
metadata_out = os.path.join(PROCESSED_DIR, "metadata.csv")
metadata_df.to_csv(metadata_out, index=False)
```

**DEA Results**

```python
Comparisons = [
  "CNR401_vs_control", 
  "CFA_vs_control"
  ]

column_mapping = {
  'feature_id': 'gene_id',
  'log2FoldChange_shrunk': 'LFC_shrunk',
  'log2FoldChange': 'LFC_raw',
  'lfcSE': 'lfcSE_shrunk'
}

column_order = ['gene_id', 'baseMean', 'LFC_shrunk', 'lfcSE_shrunk', 'stat', 
                'pvalue', 'padj', 'LFC_raw']

for comparison in Comparisons: 
  file_path = os.path.join(RAW_DIR, f"DEG_{comparison}.csv")
  df = pd.read_csv(file_path)
  df = df.rename(columns=column_mapping)
  df = df[column_order]
  out_path = os.path.join(DEA_DIR, f"DEG_{comparison}.csv" )
  df.to_csv(out_path, index = False)


df
```
